```python
import pandas as pd
import numpy as np
import tensorflow as tf
from tensorflow import keras
from sklearn.model_selection import train_test_split
```

```python
# test and train split
alz_file = pd.read_csv('12_months_joint.csv')
X = alz_file.drop(['cov'], axis = 1)
Y = alz_file['cov']
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
```

```python
# single layer perceptron implementation
model = keras.Sequential([keras.layers.Dense(1, input_dim=X_train.shape[1], activation='sigmoid')])
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
model.fit(X_train, Y_train, epochs=250, batch_size=32, verbose=0)

loss, accuracy = model.evaluate(X_test, Y_test)
print(f'Accuracy: {accuracy}')
```
